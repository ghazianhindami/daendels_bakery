```python
import pandas as pd
import re
```

```python
from sqlalchemy import create_engine
```

```python
DB_USER ="admin"
DB_PASS ="admin123"
DB_HOST ="localhost"
DB_PORT ="5432"
DB_NAME ="project_roti"
```

```python
engine = create_engine(f"postgresql+psycopg2://{DB_USER}:{DB_PASS}@{DB_HOST}/{DB_NAME}")
```

```python
import os

folder_path = r"C:\Users\ruuke\Documents\Project\Project Roti\bakery_daendels_data_csv"
files = [f for f in os.listdir(folder_path) if f.endswith('.csv')]

for file in files:
    filex = os.path.join(folder_path, file)
    
    table_name = re.sub(r'^\d+_', '', file).replace(".csv", "")
    print(f"\n===== {file} =====")
    df = pd.read_csv(filex)
    delete_statement = f"TRUNCATE {table_name} CASCADE;"
    with engine.connect() as connection:
        connection.execute(delete_statement)    

    df.to_sql(table_name, engine, if_exists='append', index=False)
    print(table_name, "uploaded successfully.")
```
