```python
from sqlalchemy import create_engine
```

```python
DB_USER ="admin"
DB_PASS ="admin123"
DB_HOST ="localhost"
DB_PORT ="5432"
DB_NAME ="project_roti_partition"
```

```python
engine = create_engine(f"postgresql+psycopg2://{DB_USER}:{DB_PASS}@{DB_HOST}/{DB_NAME}")
```

```python
y = 2029
l = y
for i in range(1,13):
    k = i + 1
    if k == 13:
        k = 1
        l = l + 1
    partition_statement = f''' DROP TABLE IF EXISTS order_y{y}m{i};
    CREATE TABLE orders_y{y}m{i} PARTITION OF orders
    FOR VALUES FROM ('{y}-{i}-01') TO ('{l}-{k}-01');'''
    with engine.connect() as connection:
        connection.execute(partition_statement) 
    print(partition_statement)  
```
